```python
import pandas as pd
from sqlalchemy import create_engine

file_path = r'C:\Users\marko\Desktop\Kaggle_real_estate_unzziped\Kaggle real estate\Real_Estate_Sales_2001-2020_GL.csv'

df = pd.read_csv(file_path)

print("✅ CSV je uspješno učitan! Broj redaka i stupaca:", df.shape)

DB_PASSWORD = "sql"  
engine = create_engine(f'postgresql://postgres:{DB_PASSWORD}@localhost:5432/real_estate_db')

df.to_sql('real_estate', engine, if_exists='replace', index=False)

print("✅ Podaci su uspješno uvezeni u PostgreSQL tablicu 'real_estate'!")
```
